```python
import numpy as np
# Sigmoid activation function
def sigmoid(z):
    return 1 / (1 + np.exp(-z))
    # STEP 1: INPUT LAYER
X = np.array([0.70, 0.80, 0.30, 0.40, 0.50, 0.60])


# STEP 2: HIDDEN LAYER 1
W1 = np.array([[ 1.2,  1.5, -1.0,  0.2,  0.5, -0.3],
    [-0.8, -0.6,  1.4, -0.1, -0.2,  1.1],
    [ 0.3,  0.4, -0.3,  0.9,  1.2, -0.2]
])
b1 = np.array([-1.0, 0.2, -0.8])
print("INPUT LAYER")
print("X =", X)
print("HIDDEN LAYER")
print("H =", W1)
print("INPUT LAYER")
print("B =", b1)
Z1 = W1 @ X + b1
A1 = sigmoid(Z1)

print("\nHIDDEN LAYER 1")
for i in range(3):
    print(f"H{i+1}: Z = {Z1[i]:.4f}, Sigmoid(Z) = {A1[i]:.4f}")

# STEP 3: OUTPUT NEURON
W2 = np.array([-2.2, 2.0, -1.0])
b2 = -0.1
Z_out = W2 @ A1 + b2

print("\nOUTPUT NEURON")
print(f"Output Z = {Z_out:.4f}")

# STEP 4: OUTPUT SIGMOID
P = sigmoid(Z_out)
print("\nOUTPUT SIGMOID")
print(f"Default probability = {P:.4f}")
print(f"Default probability = {P*100:.2f}%")

# STEP 5: CLASSIFICATION
prediction = 1 if P >= 0.5 else 0

print("\nFINAL PREDICTION")
print("Class =", prediction)
print("Result =", "DEFAULT" if prediction == 1 else "NO DEFAULT")
```
